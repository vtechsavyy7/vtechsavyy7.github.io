## Uniform Random Variable

A uniform random variable is a type of continuous random variable that has an equal probability of taking any value within a specified range. 
It is defined by two parameters: the minimum value (a) and the maximum value (b). 

- The probability density function (PDF) of a uniform random variable is constant over the interval [a, b] and zero outside this interval. It is given by the formula:

$$
f(x) = \begin{cases}
\frac{1}{b - a} & \text{for } a \leq x \leq b \\
0 & \text{otherwise}
\end{cases}
$$

- The expected value (mean) of a uniform random variable can be derived from its PDF and is calculated as follows:

$$
E[X] = \int_{a}^{b} x f(x) dx = \int_{a}^{b} x \cdot \frac{1}{b - a} dx = \frac{1}{b - a} \int_{a}^{b} x dx = \frac{1}{b - a} \left[ \frac{x^2}{2} \right]_{a}^{b} = \frac{1}{b - a} \left( \frac{b^2 - a^2}{2} \right) = \frac{a + b}{2}
$$    


```python
# Plot the PDF of a continous uniform random variable

x = np.linspace(-5, 5, 1000)
y = np.where((x >= -1) & (x <= 1), 0.5, 0)
plt.plot(x, y)
plt.title("PDF of Uniform Random Variable")
plt.xlabel("x")
plt.ylabel("Probability Density")
plt.grid()
plt.show()
```

## Exponential Random Variable

- An exponential random variable is a type of continuous random variable that models the time between events in a Poisson process. It is defined by a single parameter, the rate parameter (λ), which represents the average number of events per unit time.

- The probability density function (PDF) of an exponential random variable is given by the formula:

$$
f(x) = \begin{cases}
\lambda e^{-\lambda x} & \text{for } x \geq 0 \\
0 & \text{otherwise}
\end{cases}
