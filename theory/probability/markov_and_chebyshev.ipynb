## Chebyshev Inequality

- The Chebyshev inequality provides a bound on the probability that a random variable deviates from its mean by more than a certain amount.
- It is applicable to any random variable with a finite mean and variance.
- The bound value depends on the variance of the random variable and the distance from the mean.

Formally, it can be stated as follows: Given a random variable $X$ with mean $\mu$ and variance $\sigma^2$, for any $a > 0$:

$$
\Pr(|X - \mu| \geq a) \leq \frac{\sigma^2}{a^2}
$$

### Proof of Chebyshev's Inequality:
- Let $D^2 = (X - \mu)^2$. Then, $D^2$ is a non-negative random variable$.
- By Markov's inequality, we have:
$$
\Pr(D^2 \geq a^2) \leq \frac{\mathbb{E}[D^2]}{a^2} = \frac{\mathbb{E}[(X - \mu)^2]}{a^2} = \frac{\sigma^2}{a^2}
$$
- Since $D^2 \geq a^2$ is an equivalent event to $|X - \mu| \geq a$, we have:
$$
\Pr(|X - \mu| \geq a) \leq \frac{\sigma^2}{a^2}
$$

```python
## Plot the Chebyshev bounds for a normal distribution

mu = 2.0
sigma = 0.45

x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
pdf_X = (1/(sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu)/sigma)**2)

# Define the Chebyshev bounds for different a values
a_values = [1, 2, 3]
a_colors = ['red', 'green', 'orange']
chebyshev_bounds = [sigma**2 / a**2 for a in a_values]

# Plot the PDF of X
plt.figure(figsize=(10, 6))
plt.plot(x, pdf_X, label='PDF of X', color='blue')
# Plot the Chebyshev bounds
for a, bound in zip(a_values, chebyshev_bounds):
    plt.axvline(mu - a*sigma, color=a_colors[a_values.index(a)], linestyle='--', label=f'Bound for (a={a} => {bound:.2f})')
    plt.axvline(mu + a*sigma, color=a_colors[a_values.index(a)], linestyle='--', label=f'Bound for (a={a} => {bound:.2f})')
plt.title('PDF of Normal Distribution and Chebyshev Bounds')
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.legend()
plt.grid()
plt.show()

# Separately, we can also plot the Chebyshev bounds as a function of a
# for 4 different sigma values (0.1, 0.5, 1, 2) to visualize how the bounds change with a and sigma.
 
a_range = np.linspace(0.1, 5, 100)
chebyshev_bounds_range = {sigma_val: sigma_val**2 / a_range**2 for sigma_val in [0.1, 0.5, 1, 2]}

plt.figure(figsize=(10, 6))
for sigma_val, bounds in chebyshev_bounds_range.items():
    plt.plot(a_range, bounds, label=f'Chebyshev Bound (σ={sigma_val})')
plt.title('Chebyshev Bound as a Function of a')
plt.xlabel('a')
plt.ylabel('Chebyshev Bound (σ² / a²)')
plt.axhline(0, color='black', linestyle='--')
plt.axvline(0, color='black', linestyle='--')
plt.legend()
plt.grid()
plt.show()
```

### Note:
- The bound provided by Chebyshev's inequality is not tight in general, but it is useful for providing a conservative estimate of the probability of deviation from the mean.
- For exampple for deviations less than 1 standard deviation, the bound is not very informative. In fact, it can be counterintuitive, as it gives an upper bond of greater than 1, which is not possible for a probability. However, for larger deviations, the bound becomes more useful.
